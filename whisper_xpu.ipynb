```python
import whisper
import torch

```

```python
torch.xpu.is_available()
```

```python
torch.xpu.init()
```

```python
torch.xpu.set_device(0)
```

```python
torch.xpu.is_initialized()
```

```python
torch.xpu.is_bf16_supported()
```

```python
file = input("Pon el nombre del archivo que quieras pasar a txt: ")
out_file = input("Cómo quieres que se llame el archivo de salida? añade .txt: ")


device = torch.device('xpu')
model = whisper.load_model("turbo", device=device) #aquí donde dice turbo cambiale según el modelo que quieras

result = model.transcribe(file, fp16=True) #fp16 False si no hay gpu 
with open(out_file, "w", encoding = "utf8") as f:
    f.write(result["text"])
print("terminado y guardado") 
```

```python
torch.xpu.empty_cache()
```
